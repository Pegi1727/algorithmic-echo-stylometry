# Figure Replication Notebook

Replicates the study figures from `data_long.csv` / `data_wide.csv` using matplotlib/seaborn.

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', font_scale=1.1)
long_df = pd.read_csv('data_long.csv')
long_df['TimePoint'] = pd.Categorical(long_df['TimePoint'], ['T1','T2','T3'], ordered=True)
wide_df = pd.read_csv('data_wide.csv')
palette = sns.color_palette('deep', long_df['Profile'].nunique())
prof_colors = dict(zip(sorted(long_df['Profile'].unique()), palette))
```

## Figure 1: Longitudinal trajectories (individual + group mean)

```python
fig, ax = plt.subplots(figsize=(8,6))
sns.lineplot(data=long_df, x='TimePoint', y='Stylometry_Composite',
             hue='Profile', units='ParticipantID', estimator=None,
             alpha=0.2, legend=False, ax=ax)
sns.lineplot(data=long_df, x='TimePoint', y='Stylometry_Composite',
             hue='Profile', errorbar=None, linewidth=2.5, ax=ax)
ax.set(title='Longitudinal Trajectory of Stylometric Composite',
       xlabel='Time Point', ylabel='Composite Score')
plt.tight_layout()
plt.savefig('nb_fig_trajectory.png', dpi=300)
plt.show()
```

## Figure 2: Drift by profile

```python
fig, ax = plt.subplots(figsize=(7,6))
sns.boxplot(data=wide_df, x='Profile', y='Drift', palette=prof_colors, ax=ax)
sns.stripplot(data=wide_df, x='Profile', y='Drift', color='0.25',
              size=4, alpha=0.5, jitter=0.15, ax=ax)
ax.set(title='Stylometric Drift by Writing Profile', xlabel='', ylabel='Drift (T3 - T1)')
plt.tight_layout()
plt.savefig('nb_fig_drift.png', dpi=300)
plt.show()
```

## Figure 3: Means with 95% CI

```python
fig, ax = plt.subplots(figsize=(8,6))
sns.pointplot(data=long_df, x='TimePoint', y='Stylometry_Composite',
              hue='Profile', errorbar=('ci', 95), dodge=0.25, ax=ax)
ax.set(title='Mean Composite Score (95% CI) by Time Point',
       xlabel='Time Point', ylabel='Mean Score')
plt.tight_layout()
plt.savefig('nb_fig_means_ci.png', dpi=300)
plt.show()
```

## Figure 4: Distribution across waves

```python
g = sns.displot(data=long_df, x='Stylometry_Composite', hue='TimePoint',
                col='Profile', kind='kde', fill=True, palette='Blues', height=4)
g.set(title='', xlabels='Composite Score')
g.figure.suptitle('Distribution of Composite Scores Across Waves', y=1.03)
plt.savefig('nb_fig_density.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Figure 5: Correlation heatmap of wave scores

```python
fig, ax = plt.subplots(figsize=(5,4))
sns.heatmap(wide_df[['T1','T2','T3']].corr(), annot=True, cmap='vlag',
            vmin=-1, vmax=1, ax=ax)
ax.set(title='Correlation Between Waves')
plt.tight_layout()
plt.savefig('nb_fig_corr.png', dpi=300)
plt.show()
```
